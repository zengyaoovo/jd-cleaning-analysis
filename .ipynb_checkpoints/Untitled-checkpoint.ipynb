```python
import pandas as pd

# 按 UTF-8 读取数据集，清理列名前后空格并查看基本信息
comment_data = pd.read_csv('jd_comment_small.csv', encoding='utf-8')
comment_data.columns = comment_data.columns.str.strip()
print(comment_data.shape)
comment_data.head()
```

```python
comment_data.isnull().sum()
```

```python
# 将原始列名重命名为简洁字段，便于后续引用
comment_data = comment_data.rename(columns={
    '爬取时间(__time)': '爬取时间',
    '爬取链接(__url)': '爬取链接',
    '商品ID(product_id)': '商品ID',
    '评价时间(publish_time)': '评价时间戳',
    '评分（总分5分）(score)': '评分',
    '评价内容(content)': '评论内容',
    '评价者(author_name)': '评价者',
    '评价者会员等级(author_level)': '会员级别',
    '商品sku(product_sku)': '商品SKU',
    '评价标签(tags)': '评价标签'
})
print(comment_data.columns.tolist())
comment_data.head()
```

```python
# 把每一列的缺失值 NaN 填充成指定内容，部分列转为数字类型
comment_data['评论内容'] = comment_data['评论内容'].fillna('')
comment_data['会员级别'] = comment_data['会员级别'].fillna('')
comment_data['评价者'] = comment_data['评价者'].fillna('')
comment_data['商品SKU'] = comment_data['商品SKU'].fillna('')
comment_data['评价标签'] = comment_data['评价标签'].fillna('[]')
comment_data['评分'] = pd.to_numeric(comment_data['评分'], errors='coerce').fillna(0)
comment_data['商品ID'] = comment_data['商品ID'].fillna(0)
comment_data['评价时间戳'] = pd.to_numeric(comment_data['评价时间戳'], errors='coerce').fillna(0)
comment_data['爬取时间'] = comment_data['爬取时间'].fillna('')
comment_data['爬取链接'] = comment_data['爬取链接'].fillna('')

# 定义默认评价的正则模式，用于过滤无效评论
default_pattern = '未填写评价内容|默认好评|系统默认评价|未及时填写评价内容|没有填写内容'
# 标记出所有默认评价或空评论
mask_default = (
    comment_data['评论内容'].str.contains(default_pattern, na=False) |
    (comment_data['评论内容'].str.strip() == '')
)
# 删除默认评价和空评论
comment_data = comment_data[~mask_default].copy()

# 按商品ID、评价者、评论内容、评价时间戳去重
before = len(comment_data)
comment_data = comment_data.drop_duplicates(
    subset=['商品ID', '评价者', '评论内容', '评价时间戳'],
    keep='first'
)
print("去重删除数量：", before - len(comment_data))
print("有效评论数量：", len(comment_data))

# 过滤会员级别为空的数据，便于后续会员维度分析
comment_data = comment_data[comment_data['会员级别'].str.strip() != ''].copy()
print("过滤空会员级别后数据量：", len(comment_data))
```

```python
# 将 Unix 时间戳转换为北京时间，并提取年份、月份、小时
comment_data['评价时间'] = (
    pd.to_datetime(comment_data['评价时间戳'], unit='s', errors='coerce') +
    pd.Timedelta(hours=8)
)
comment_data['年份'] = comment_data['评价时间'].dt.year
comment_data['月份'] = comment_data['评价时间'].dt.month
comment_data['小时'] = comment_data['评价时间'].dt.hour

# 过滤明显异常的年份数据
comment_data = comment_data[comment_data['年份'].between(2010, 2030)].copy()
print(comment_data[['评价时间戳', '评价时间', '年份', '月份', '小时']].head())
print("时间处理后的数据量：", len(comment_data))
```

```python
from sklearn.preprocessing import LabelEncoder

# 对会员级别进行整数编码
level_encoder = LabelEncoder()
comment_data['会员级别_编码'] = level_encoder.fit_transform(comment_data['会员级别'])
#level_encoder是LabelEncoder 的实例对象，里面保存了fit之后学到的整套映射规则
userLevel = level_encoder.classes_
print("会员等级类别：", userLevel)

# 对商品SKU进行整数编码
sku_encoder = LabelEncoder()
comment_data['商品SKU_编码'] = sku_encoder.fit_transform(comment_data['商品SKU'])
phone_type = sku_encoder.classes_
print("商品SKU类别数量：", len(phone_type))

# 对商品ID进行整数编码
product_encoder = LabelEncoder()
comment_data['商品ID_编码'] = product_encoder.fit_transform(comment_data['商品ID'].astype(str))
print("商品ID类别数量：", len(product_encoder.classes_))
```

数据清洗完毕，开始数据分析流程

```python
from pyecharts.charts import Bar
from pyecharts import options as opts

# 统计每个小时的评价次数
attr = [f"{i}时" for i in range(24)]
v1 = [int((comment_data['小时'] == i).sum()) for i in range(24)]

# 创建柱状图并添加数据
bar = Bar()
bar.add_xaxis(attr)
bar.add_yaxis("评价次数", v1)
bar.set_global_opts(
    title_opts=opts.TitleOpts(title="评价时段分布", pos_left="center"),
    xaxis_opts=opts.AxisOpts(name="小时"),
    yaxis_opts=opts.AxisOpts(name="评价次数"),
)
bar.render("评价时段柱状图.html")
print("已生成：评价时段柱状图.html")
```

```python
from pyecharts.charts import Line

# 遍历每个会员等级，统计其每月评价数量
v = [[]]
for level in level_encoder.classes_:
    data = comment_data[(comment_data['会员级别'] == level) & (comment_data['月份'].notna())]
    month = data['月份'].astype(int).tolist()
    counts = [month.count(m) for m in range(1, 13)]
    v.append(counts)
    print(level, counts)

# 创建折线图，为每个会员等级添加一条折线
attr_month = ["一月", "二月", "三月", "四月", "五月", "六月",
              "七月", "八月", "九月", "十月", "十一月", "十二月"]
line = Line()
line.add_xaxis(attr_month)
for i, level in enumerate(level_encoder.classes_, start=1):
    line.add_yaxis(level, v[i])
line.set_global_opts(
    title_opts=opts.TitleOpts(title="不同会员等级每月评价数量", pos_left="center"),
    xaxis_opts=opts.AxisOpts(name="月份"),
    yaxis_opts=opts.AxisOpts(name="评价数量"),
)
line.render("会员每月评价折线图.html")
print("已生成：会员每月评价折线图.html")

# 计算各会员等级的评价占比
total = len(comment_data)
print("\n会员评价占比：")
for level in level_encoder.classes_:
    cnt = (comment_data['会员级别'] == level).sum()
    print(f"{level}所占比例{cnt / total * 100:.2f}%")
```

```python
from pyecharts.charts import Pie

# 筛选有效评分（1-5分）并统计各分数数量
score_data = comment_data[comment_data['评分'].between(1, 5)]
score_counts = score_data['评分'].value_counts().sort_index()

# 创建饼图展示评分分布
pie_score = (
    Pie()
    .add(
        "",
        [[f"{int(k)}分", int(v)] for k, v in score_counts.items()],
        radius=["40%", "75%"],
        label_opts=opts.LabelOpts(formatter="{b}: {d}%"),
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="评分分布", pos_left="center"),
        legend_opts=opts.LegendOpts(is_show=False),
    )
)
pie_score.render("评分分布.html")
print("已生成：评分分布.html")
```

```python
import json
from collections import Counter

# 遍历评价标签列，解析 JSON 并统计每个标签的出现次数
tag_counter = Counter()
for tags in comment_data['评价标签']:
    try:
        tag_list = json.loads(tags) if isinstance(tags, str) else []
        if isinstance(tag_list, list):
            tag_counter.update([str(t).strip() for t in tag_list if str(t).strip()])
    except Exception:
        pass

# 获取出现次数最多的 20 个标签
top_tags = tag_counter.most_common(20)
print("Top20 评价标签：")
for tag, cnt in top_tags:
    print(tag,cnt,end=', ')

# 如果存在标签数据，则绘制柱状图
if top_tags:
    tag_bar = Bar()
    tag_bar.add_xaxis([t[0] for t in top_tags])
    tag_bar.add_yaxis("出现次数", [t[1] for t in top_tags])
    tag_bar.set_global_opts(
        title_opts=opts.TitleOpts(title="评价标签 Top20", pos_left="center"),
        xaxis_opts=opts.AxisOpts(axislabel_opts=opts.LabelOpts(rotate=45)),
    )
    tag_bar.render("评价标签Top20.html")
    print("已生成：评价标签Top20.html")
```

```python
import re
import os
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import jieba

# 将所有评论拼接为一个长字符串，并清洗无效字符
data = comment_data['评论内容'].astype(str).tolist()
dataStr = ','.join(data)
dataStr = dataStr.replace('hellip', '')
dataStr = dataStr.replace('&hellip;', '')
dataStr = dataStr.replace('&mdash;', '')
dataStr = dataStr.replace('。', '')
dataStr = dataStr.replace('，', '')
dataStr = dataStr.replace('！', '')
dataStr = dataStr.replace(',', '')
dataStr = dataStr.replace('&', '')
dataStr = dataStr.replace('...', '')
dataStr = re.sub(r'[^\w\u4e00-\u9fa5]+', ' ', dataStr)

# 定义停用词集合，过滤无意义词汇
stopwords = set([
    '京东', '东西', '不错', '可以', '非常', '就是', '没有', '什么', '这个', '感觉',
    '使用', '收到', '买', '买给', '买了', '好评', '默认', '评价', '用户', '我们',
    '他们', '自己', '一个', '真的', '还是', '已经', '特别', '有点', '比较', '的话',
    '但是', '因为', '所以', '而且', '以及', '或者', '如果', '那个', '怎么', '为什么',
    '觉得', '快递', '物流', '服务', '质量', '价格', '商品', '产品', '手机', '很好',
    '好', '很', '是', '的', '了', '在', '有', '和', '与', '我', '你', '他', '她',
    '它', '们', '这', '那', '不', '也', '都', '就', '还', '又', '太', '更', '最',
    '会', '能', '要', '让', '给', '对', '从', '到', '把', '被', '为', '于', '上',
    '下', '中', '里', '外', '前', '后', '左', '右', '时', '分', '秒', '年', '月',
    '日', '点', '一次', '一直', '一下', '以后', '一点', '一款', '一家', '一样',
    '一般', '一起', '一定', '一切'
])

# 使用 jieba 分词，并过滤停用词和单字
wordlist = jieba.cut(dataStr, cut_all=False)
word_string = " ".join([w for w in wordlist if w not in stopwords and len(w) > 1])

# 读取背景图
mask_path = 'picture/cloud.jpg'
backgroup_Image = plt.imread(mask_path)

# 创建词云并保存图片
os.makedirs('picture', exist_ok=True)
wordcloud = WordCloud(
    font_path=r'C:\Windows\Fonts\msyh.ttc',
    background_color="white",
    mask=backgroup_Image,
    width=1000,
    height=860,
    margin=2
).generate(word_string)

plt.imshow(wordcloud)
plt.axis("off")
plt.show()
wordcloud.to_file('picture/phoneComment.jpg')
print("已生成：picture/phoneComment.jpg")
```
